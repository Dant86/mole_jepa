## Model Registration Example

```python
import dotenv
import os

from mole_jepa import config, registry

dotenv.load_dotenv()
registry_path = os.getenv("REGISTRY_PATH")
checkpoint_dir = os.getenv("CHECKPOINT_DIR")
```

```python
base_cfg = config.ModelConfig(
    freeze_image_encoder=False,
    jepa_regularize_z_i=True,
)

_ = registry.register(
    "base_molejepa",
    base_cfg,
    checkpoint_dir=checkpoint_dir,
    registry_dir=registry_path,
    overwrite=True,
)
```

```python
base_cfg = config.ModelConfig(
    freeze_image_encoder=True,
    jepa_regularize_z_i=False,
)

_ = registry.register(
    "base_molejepa_frozen",
    base_cfg,
    checkpoint_dir=checkpoint_dir,
    registry_dir=registry_path,
    overwrite=True,
)
```

```python
base_cfg = config.ModelConfig(
    freeze_image_encoder=True,
    contrastive=True,
)

_ = registry.register(
    "base_molejepa_contrastive",
    base_cfg,
    checkpoint_dir=checkpoint_dir,
    registry_dir=registry_path,
    overwrite=True,
)
```
